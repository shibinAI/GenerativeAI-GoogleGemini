```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
import numpy as np

# Create dummy data with categorical features
dummy_data = pd.DataFrame({
    'Color': ['Red', 'Blue', 'Green', 'Red', 'Blue', 'Green'],
    'Size': ['Small', 'Medium', 'Large', 'Small', 'Large', 'Medium'],
    'Brand': ['A', 'B', 'A', 'C', 'B', 'C'],
    'Price': [100, 200, 150, 120, 250, 180]
})

print("Original Data:")
print(dummy_data)
print("\n" + "="*50 + "\n")

# Method 1: Using pandas get_dummies (simple approach)
encoded_data_pandas = pd.get_dummies(dummy_data, columns=['Color', 'Size', 'Brand'], drop_first=False)
print("One-Hot Encoded Data (using pandas get_dummies):")
print(encoded_data_pandas)
print("\n" + "="*50 + "\n")

# Method 2: Using scikit-learn OneHotEncoder (more scalable)
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoded_features = encoder.fit_transform(dummy_data[['Color', 'Size', 'Brand']])
encoded_column_names = encoder.get_feature_names_out(['Color', 'Size', 'Brand'])

encoded_data_sklearn = pd.DataFrame(encoded_features, columns=encoded_column_names)
# Add the Price column back
encoded_data_sklearn['Price'] = dummy_data['Price'].values

print("One-Hot Encoded Data (using scikit-learn):")
print(encoded_data_sklearn)
```
